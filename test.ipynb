```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from model import llm
from langchain_core.output_parsers import StrOutputParser
from prompt import filter_prompt, note_prompt
import time
```

```python
from Video_id_generater import get_id
from Transcript import get_transcripts
```

```python
video_id = get_id("https://youtu.be/pJdMxwXBsk0")
trans = get_transcripts(video_id)
```

```python

def get_chunks(transcript):
    splitter = RecursiveCharacterTextSplitter(
        chunk_size = 2000,
        chunk_overlap = 0,
    )
    return splitter.create_documents([transcript])


chain = filter_prompt | llm | StrOutputParser()
note_chain = note_prompt | llm | StrOutputParser()


def filter_chunks(chunks):
    filtered_chunks = []

    for chunk in chunks:
        result = chain.invoke({"chunk": chunk})
        filtered_chunks.append(result)
        time.sleep(5)

    final_transcript = "\n\n".join(filtered_chunks)

    return final_transcript



def notes_maker(chunks):
    filtered_chunks = []

    for chunk in chunks:
        result = note_chain.invoke({"transcript": chunk})
        filtered_chunks.append(result)
        time.sleep(5)

    final_notes = "\n\n".join(filtered_chunks)

    return final_notes
```
